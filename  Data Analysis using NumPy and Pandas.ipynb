# Data Analysis using NumPy and Pandas

### To learn how to use NumPy for creating arrays, performing basic array operations, and conducting numerical computations through analyzing temperature data. 

### Numpy Array Operations:

```python
# Given a list of daily average temperatures (in degrees Celsius)recorded over two weeks.Using Numpy to analyze the data
```

```python
import numpy as np
import pandas as pd
```

```python
# Create a 1D NumPy Array:
temperatures_w1=np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
```

```python
# Inspection and Properties:
print('Shape Of The Array:',temperatures_w1.shape)
print('Data Type Of Array:',temperatures_w1.dtype)
print('Number Of Elements In The Array:',temperatures_w1.size)
```

```python
# Array Operations: Convert the temperatures_w1 from Celsius to Fahrenheit
fahrenheit=(temperatures_w1*(9/5))+32
print('Fahrenheit:',fahrenheit)

# Maximum, Minimum, and Mean temperatures for the week.
print(f'Maximum Temperature:{np.max(temperatures_w1)}\nMinimum Temperature:{np.min(temperatures_w1)}\nMean Temperature:{np.mean(temperatures_w1)}')
```

```python
# Array Slicing and Indexing:

# Temperatures for the first three days of the week.
print('Temperatures for the first three days of the week:',temperatures_w1[0:3])

# Temperatures for the weekend (last two days).
print('Temperatures for the weekend (last two days):',temperatures_w1[-2:])

#Temperatures for the middle three days of the week.
print('Temperatures for the middle three days of the week:',temperatures_w1[-5:-2])
```

```python
# Create a 2D Array:
#Create a 2D NumPy array named temperatures where each row represents temperatures for a different week:
temperatures=np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],[19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
print(f'Week1:{temperatures[0]}\nWeek2:{temperatures[1]}')
```

```python
#Inspect and Slice the 2D Array:
print('Dimension Of Array:',temperatures.ndim)
print('Shape Of The Array:',temperatures.shape)
print('Data Type Of Array:',temperatures.dtype)
print('Number Of Elements In The Array:',temperatures.size)
```

```python
#Temperatures for each week and the weekends (last two days) for both weeks.
print(f'Week1:{temperatures[0]}\nWeek2:{temperatures[1]}')
print(f'Last two days Week1:{temperatures[0][-2:]}\nLast two days Week2:{temperatures[1,-2:]}')
```

# Pandas

```python
# Creating Pandas Series:
marks=pd.Series(data=[95, 92, 89, 85, 80],index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)
```

```python
# Indexing and Slicing:
# Mark of the 1st rank student
print('Mark of the 1st rank student:',marks['Rank1'])
print()

# Retrieve the marks of the top 3 ranks by specifying their index labels.
print('Marks of the top 3 ranks:\n',marks.loc[['Rank1','Rank2','Rank3']])
print()

# Retrieve the mark of the 3rd rank student.
print('Mark of the 3rd rank student:',marks.iloc[2])
print()

# Filter the Series and retrieve the ranks where the marks are greater than 90.
print('Retrieve the ranks where the marks are greater than 90:\n',marks[marks>90])
```

```python
#Manipulating Series:
# Modify the mark of the student with the 1st rank to 100.
marks.loc['Rank1']=100
print(marks)
print()

#Remove the entry corresponding to the last rank from the Series.
marks=marks.drop(marks.index[-1])
print(marks)
print()

#Compute the CGPA by dividing each mark in the Series by 10.
print('CGPA')
cgpa=marks/10
print(cgpa)
```

# Pandas DataFrame:

```python
dict1={'TransactionID':[101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
       'ProductCategory':['Electronics', 'Clothing', 'Electronics', 'Furniture','Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
       'Region':['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West','South', 'North'],
       'Amount':[200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
      }
dict1
```

```python
transactions=pd.DataFrame(dict1)
print(transactions)
```

```python
#Data Exploration:
print(transactions)
print()
print('Head:\n',transactions.head())
print()
print('Tail:\n',transactions.tail())
```

```python
print('Shape:',transactions.shape)
```

```python
print('Data Type:\n',transactions.dtypes)
```

```python
print('Columns\n',transactions.columns)
```

```python
print(transactions.info())
```

```python
# 'ProductCategory' and 'Amount' columns.
transactions[['ProductCategory','Amount']]
```

```python
# The last 3 columns of the DataFrame using iloc or loc.
transactions.iloc[:,-3:]
```

```python
# The 'Region' is 'North' and 'Amount' is greater than 200.
transactions[(transactions['Region'] =='North') & (transactions['Amount']>200)]
```

```python
# Value counts for the 'ProductCategory' column.
print(transactions.value_counts('ProductCategory'))
```

```python
# Unique values in the 'Region' column.
print(transactions['Region'].unique())
```

```python
# Group by 'Region' and find the mean amount for each region.
print(transactions.groupby('Region')['Amount'].mean())
```

```python
# Manipulating the DataFrame:
# Modify the 'Amount' for the transaction with TransactionID 102 to 165.
transactions.loc[transactions['TransactionID']==102,'Amount']=165
transactions
```

```python
# Add a new column 'Discount' by calculating 10% of the 'Amount'.
transactions['Discount']=(10/100)*transactions['Amount']
transactions
```

```python
# Remove the row with TransactionID 109.
transactions = transactions.drop(transactions[transactions['TransactionID'] == 109].index)
transactions
```

```python
# Delete the 'Discount' column from the DataFrame.
transactions=transactions.drop('Discount',axis=1)
transactions
```
